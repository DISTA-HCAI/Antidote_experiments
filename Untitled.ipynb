```python
def preprocess_for_dpo(
    examples: Dict[str, List[str]], 
    tokenizer: AutoTokenizer, 
    max_length: int = 1024,
    label_pad_token_id: int = -100
) -> Dict[str, torch.Tensor]:
    """
    Preprocesses a batch of preference data for DPO training.

    This function takes a dictionary of lists (as provided by `dataset.map(batched=True)`),
    where each list corresponds to 'prompt', 'chosen', and 'rejected' strings.
    It formats and tokenizes the data into the six required tensors for the custom DPO loss function.

    Args:
        examples (Dict[str, List[str]]): A batch of examples from a Hugging Face dataset.
                                         Expected keys: 'prompt', 'chosen', 'rejected'.
        tokenizer (AutoTokenizer): The tokenizer for the model.
        max_length (int): The maximum sequence length for truncation.
        label_pad_token_id (int): The ID to use for masking out tokens in the labels.

    Returns:
        Dict[str, torch.Tensor]: A dictionary containing the six required tokenized tensors.
    """
    # Ensure the tokenizer has a pad token; for decoder-only models, this is often the EOS token.
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    batch_size = len(examples['prompt'])
    
    # --- 1. Format chosen and rejected completions using the chat template ---
    chosen_full_texts = []
    rejected_full_texts = []
    prompt_only_texts = []

    for i in range(batch_size):
        prompt_text = examples['prompt'][i]
        chosen_text = examples['chosen'][i]
        rejected_text = examples['rejected'][i]

        # Format for chosen response
        messages_chosen = [
            {"role": "user", "content": prompt_text},
            {"role": "assistant", "content": chosen_text}
        ]
        chosen_full_texts.append(tokenizer.apply_chat_template(messages_chosen, tokenize=False))
        
        # Format for rejected response
        messages_rejected = [
            {"role": "user", "content": prompt_text},
            {"role": "assistant", "content": rejected_text}
        ]
        rejected_full_texts.append(tokenizer.apply_chat_template(messages_rejected, tokenize=False))
        
        # Format prompt only to calculate its length for masking
        messages_prompt = [{"role": "user", "content": prompt_text}]
        prompt_only_texts.append(tokenizer.apply_chat_template(messages_prompt, tokenize=False, add_generation_prompt=True))

    # --- 2. Tokenize all formatted texts ---
    tokenized_chosen = tokenizer(
        chosen_full_texts,
        truncation=True,
        max_length=max_length,
        padding="longest" # Pad to the longest sequence in this batch
    )
    tokenized_rejected = tokenizer(
        rejected_full_texts,
        truncation=True,
        max_length=max_length,
        padding="longest"
    )
    # Tokenize prompts without padding to get their true lengths
    tokenized_prompts = tokenizer(prompt_only_texts, truncation=True, max_length=max_length)

    # --- 3. Create the final batch dictionary ---
    batch_dict = {}

    # --- 4. Create Labels by masking out the prompt portion ---
    # The labels are a copy of the input_ids, with prompt tokens set to -100
    chosen_labels = torch.tensor(tokenized_chosen['input_ids']).clone()
    rejected_labels = torch.tensor(tokenized_rejected['input_ids']).clone()
    
    for i in range(batch_size):
        prompt_len = len(tokenized_prompts['input_ids'][i])
        
        # Mask chosen labels
        chosen_labels[i, :prompt_len] = label_pad_token_id
        # Also mask padding tokens if any exist in the response part
        chosen_labels[i][tokenized_chosen['attention_mask'][i] == 0] = label_pad_token_id
        
        # Mask rejected labels
        rejected_labels[i, :prompt_len] = label_pad_token_id
        # Also mask padding tokens
        rejected_labels[i][tokenized_rejected['attention_mask'][i] == 0] = label_pad_token_id

    # Populate the final dictionary with all required tensors
    batch_dict["chosen_input_ids"] = torch.tensor(tokenized_chosen['input_ids'])
    batch_dict["chosen_attention_mask"] = torch.tensor(tokenized_chosen['attention_mask'])
    batch_dict["chosen_labels"] = chosen_labels

    batch_dict["rejected_input_ids"] = torch.tensor(tokenized_rejected['input_ids'])
    batch_dict["rejected_attention_mask"] = torch.tensor(tokenized_rejected['attention_mask'])
    batch_dict["rejected_labels"] = rejected_labels
    
    return batch_dict



def preprocess_for_it(
    examples: Dict[str, List[str]], 
    tokenizer: AutoTokenizer, 
    max_length: int = 1024,
    label_pad_token_id: int = -100
) -> Dict[str, torch.Tensor]:
    """
    Preprocesses a batch of instruction-tuning data for the retain loss.

    This function takes a dictionary of lists (as provided by `dataset.map(batched=True)`),
    where each list corresponds to 'prompt' and 'response' strings.
    It formats and tokenizes the data into the three required tensors for the 
    `obj_standard_max_next_token` loss function.

    Args:
        examples (Dict[str, List[str]]): A batch of examples from a Hugging Face dataset.
                                         Expected keys: 'prompt', 'response'.
        tokenizer (AutoTokenizer): The tokenizer for the model.
        max_length (int): The maximum sequence length for truncation.
        label_pad_token_id (int): The ID to use for masking out tokens in the labels.

    Returns:
        Dict[str, torch.Tensor]: A dictionary containing 'input_ids', 'attention_mask', and 'labels'.
    """
    # Ensure the tokenizer has a pad token.
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
        
    batch_size = len(examples['prompt'])
    full_texts = []
    prompt_only_texts = []

    # Correctly iterate over the batched examples
    for prompt, response in zip(examples['prompt'], examples['response']):
        # Format the full conversation
        messages_complete = [
            {"role": "user", "content": prompt},
            {"role": "assistant", "content": response}
        ]
        full_texts.append(tokenizer.apply_chat_template(messages_complete, tokenize=False))
        
        # Format the prompt only to get its length for masking
        messages_prompt = [{"role": "user", "content": prompt}]
        prompt_only_texts.append(tokenizer.apply_chat_template(messages_prompt, tokenize=False, add_generation_prompt=True))

    # Tokenize the full conversation texts, returning PyTorch tensors
    tokenized_complete = tokenizer(
        full_texts,
        truncation=True,
        max_length=max_length,
        padding="longest",
        return_tensors="pt"
    )

    # Tokenize the prompts without padding to get their true lengths
    tokenized_prompts = tokenizer(
        prompt_only_texts,
        truncation=True,
        max_length=max_length,
        padding=False # More efficient
    )

    # Create labels by cloning input_ids
    labels = tokenized_complete.input_ids.clone()

    for i in range(batch_size):
        prompt_len = len(tokenized_prompts.input_ids[i])
        
        # Mask the prompt portion of the labels
        labels[i, :prompt_len] = label_pad_token_id
    
    # Also mask any padding tokens that might be in the response part
    # This is implicitly handled by CrossEntropyLoss's ignore_index, but explicit is good.
    # We can rely on the attention mask for this.
    labels[tokenized_complete.attention_mask == 0] = label_pad_token_id
    
    return {
        "input_ids": tokenized_complete.input_ids,
        "attention_mask": tokenized_complete.attention_mask,
        "labels": labels,
    }    
```

## Adversarial Immunisation (Bi-Level Training)

Everything above only prepared data. This section actually runs the AntiDote
tamper-resistance procedure: an **adversary hypernetwork** watches the defender's
internal activations on a harmful prompt and learns to synthesize a malicious LoRA
patch; the **defender** (`model`'s `defender` adapter) is trained so that, even with
that patch injected, it still refuses the harmful request and preserves its general
capabilities on benign instructions.

### Fixes vs. `training.py`'s `train_tamper_resistant_model`

Porting the loop to run correctly (and on CPU) required fixing a few real bugs in the original script:

1. **Dead/leaking gradients** — the original never froze the defender's params during the adversary phase (or vice versa), and never called `optimizer_d.zero_grad()` during that phase either, so stale defender gradients from the adversary phase silently got applied on the first defender step of the next block. Fixed by explicitly toggling `requires_grad` per phase and zeroing both optimizers every step.
2. **Hook leak** — `with act_cache:` was re-entered every step, and `ActivationCache.__exit__` is a no-op, so forward hooks accumulated without bound across steps. Fixed by registering hooks once and only clearing cached values per step.
3. **Missing preprocessing / wrong keys** — the defender phase read `harmful_batch['safe_input_ids']` / `['harmful_input_ids']`, fields that don't exist in `dpo_data.json` (only `prompt`/`chosen`/`rejected`), and used the benign batch's raw prompt/response strings directly as model kwargs. Fixed by running both batches through `preprocess_for_dpo`/`preprocess_for_it` first.
4. **Chosen/rejected orientation** — in this dataset `chosen` = the harmful response and `rejected` = the safe one (see `dataset_generation.py`), which is what the *adversary* wants to prefer. The defender's safety loss needs the opposite orientation; the original code's swap logic ended up reversing it back to the adversary's orientation. Fixed by explicitly building the defender's DPO batch with `chosen`/`rejected` swapped relative to the raw dataset batch.
5. **bf16 mixed precision / gradient checkpointing / multi-GPU `Accelerator`** — all GPU-oriented; dropped for a plain single-process CPU loop.

Three more bugs turned up while actually running this loop (not just reading it), all inside `run_immunisation` itself:

6. **The adversary never learned anything, for the entire run** — `make_adversarial_patch`'s body, *including the call to the adversary hypernetwork itself*, was wrapped in `torch.no_grad()`. That means `U_gen, V_gen = adversary(...)` never built a gradient graph back to `adversary.parameters()`, so `loss_a.backward()` produced a gradient of exactly zero for the adversary on every single step, and `optimizer_a.step()` was a complete no-op. The adversary sat at its random initialization for the whole run while `history["adversary_loss"]` looked like it was training normally. Fixed by keeping only the *observation* forward pass (which just reads activations off the current, frozen defended model) under `no_grad()`, and calling the adversary itself with gradients enabled.
7. **`model.set_adapter("defender")` called on every step of both phases** — `set_adapter` re-enables `requires_grad` on the named adapter's parameters as a side effect. Calling it inside `make_adversarial_patch` (used by both phases) silently undid the `requires_grad = False` freeze applied to the defender at the top of Phase 1, so `loss_a.backward()` was also dumping huge (~1e18-scale), never-applied gradients into the defender's parameters during the adversary's own training phase. Fixed by activating the adapter once, before the block loop, instead of on every patch.
8. **Non-finite gradients silently corrupting the weights** — this loss (DPO on *summed*, not averaged, log-probabilities) is large enough in scale that the adversary's gradients occasionally overflow to `inf`. `clip_grad_norm_` with an `inf` norm computes a scaling factor of `max_norm / inf = 0`, and `0 * inf = NaN` — which `optimizer.step()` then writes straight into the weights, corrupting them for the rest of the run with no error or warning. Fixed by skipping (and printing a warning for) any optimizer step whose gradient norm isn't finite.

Also, separately: `MAX_LEN = 48` (two cells below) truncates away the *entire* response for most rows of `instruction_tuning_data.json` — with the first `N_BENIGN=8` rows selected below, 7 of the 8 end up with zero non-masked (`-100`) label tokens after truncation, so `CrossEntropyLoss` returns NaN for them and that NaN poisons `total_loss_d` (and every defender gradient after it) starting from the very first Phase-2 step. Raised to 256.

### Scale note

The paper's actual run uses `k_steps=2000`, 3 epochs, batch size 2, over the full 15,580/32,774-example datasets, on 3×A6000 GPUs. That's infeasible here. Below runs the *exact same algorithm* on a tiny slice, purely to prove the bi-level loop is wired correctly end-to-end on CPU. Scale `N_HARMFUL`/`N_BENIGN`/`k_steps`/`num_blocks`/`epochs` up once you have a GPU.

```python
def run_immunisation(
    model, adversary, tokenizer, harmful_dataloader, benign_dataloader,
    k_steps=1, num_blocks=1, epochs=1,
    lr_defender=3e-5, lr_adversary=2e-7,
    max_length=48, device=torch.device('cpu'),
):
    """
    Bi-level adversarial immunisation loop (AntiDote), adapted from
    `train_tamper_resistant_model` in training.py for a single CPU process.

    Each block alternates:
      Phase 1 - the adversary hypernetwork watches the defender's activations on a
                harmful prompt and learns to synthesize a LoRA patch that, once injected,
                makes the model prefer the harmful completion (chosen) over the safe one
                (rejected) -- simulating a malicious fine-tuner.
      Phase 2 - the defender's own LoRA ('defender' adapter) is updated so that, even with
                that same adversarial patch injected, it still prefers the safe completion,
                while also staying close to its original behaviour on benign instructions
                (LM loss + KL retention loss against the frozen reference model).
    """
    model.to(device)
    adversary.to(device)

    with torch.no_grad():
        ref_model = copy.deepcopy(model.base_model)

    # Non ottimizzerò ref_model. 
    ref_model.eval().to(device)
    for p in ref_model.parameters():
        p.requires_grad = False


    defender_params = [p for n, p in model.named_parameters() if "defender" in n]

    optimizer_d = AdamW(defender_params, lr=lr_defender)

    optimizer_a = AdamW(adversary.parameters(), lr=lr_adversary)


    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj"]
    
    act_cache = ActivationCache(model, target_modules)
    act_cache.register_hooks()

    # BUG FIX: `model.set_adapter("defender")` used to be called inside
    # `make_adversarial_patch` below, i.e. on EVERY step of BOTH phases. In PEFT,
    # `set_adapter` re-enables `requires_grad` on the named adapter's parameters as a
    # side effect -- so every call during Phase 1 silently undid the
    # `requires_grad = False` freeze applied to the defender a few lines below, and
    # `loss_a.backward()` ended up dumping huge (~1e18-scale), never-applied gradients
    # into the defender's parameters during the adversary's own training phase. There's
    # only one adapter in play here, so activating it once, up front, is enough.
    model.set_adapter("defender")

    def make_adversarial_patch(input_ids, attention_mask):

        act_cache.clear_cache()

        # BUG FIX: this used to wrap the entire function body -- including the call to
        # `adversary(...)` below -- in `torch.no_grad()`. That meant `U_gen`/`V_gen`
        # never carried a gradient graph back to `adversary.parameters()`, so
        # `loss_a.backward()` produced a gradient of EXACTLY ZERO for the adversary on
        # every single step, and `optimizer_a.step()` was a complete no-op for the whole
        # run: the adversary never moved from its random initialization, even though
        # `history["adversary_loss"]` looked like it was training normally. Only the
        # *observation* forward pass below (which just reads activations off the
        # current, frozen defended model) needs to be gradient-free; the adversary
        # itself must be called with gradients enabled so it can actually learn.
        with torch.no_grad():
            _ = model(input_ids=input_ids, attention_mask=attention_mask)

        lora_weights = {}
        for layer_name, activations in act_cache.activations.items():
            config_name = layer_name.split(".")[-1]
            U_gen, V_gen = adversary(activations.to(device), config_name)
            lora_weights[layer_name] = (U_gen, V_gen)
        return lora_weights

    def step_or_skip(params, optimizer, max_norm=1.0):
        """Clip the gradient and step -- unless the grad norm isn't finite.

        BUG FIX: the DPO loss here is computed on SUMMED (not averaged) log-probabilities,
        which makes it large enough in scale that the adversary's gradients occasionally
        overflow to `inf`. `clip_grad_norm_` with an `inf` norm computes a scaling factor
        of `max_norm / inf = 0`, and `0 * inf = NaN` -- which `optimizer.step()` then
        writes straight into the weights, silently corrupting them (no error, no
        warning) for the rest of the run. Skipping the step when this happens is cheap
        insurance against that.
        """
        grad_norm = torch.nn.utils.clip_grad_norm_(params, max_norm)
        if not torch.isfinite(grad_norm):
            print(f"    !! non-finite grad norm ({grad_norm}) -- skipping this optimizer step")
            optimizer.zero_grad()
            return grad_norm
        optimizer.step()
        return grad_norm

    harmful_iterator = iter(harmful_dataloader)
    benign_iterator = iter(benign_dataloader)
    history = {"adversary_loss": [], "defender_safety_loss": [], "defender_lm_loss": [], "defender_kl_loss": []}

    try:
        for epoch in range(epochs):
            print(f"\n===== Epoch {epoch + 1}/{epochs} =====")

            for block_idx in range(num_blocks):
                # ================= PHASE 1: TRAIN ADVERSARY =================
                print(f"--- Block {block_idx + 1}/{num_blocks}: training adversary for {k_steps} step(s) ---")
                adversary.train()

                ###### Non ottimizzare model per ora.
                model.eval()
                for _, p in model.named_parameters():
                    p.requires_grad = False
                #######

                for p in adversary.parameters():
                    p.requires_grad = True

                for step in range(k_steps):
                    try:
                        harmful_batch = next(harmful_iterator)
                    except StopIteration:
                        harmful_iterator = iter(harmful_dataloader)
                        harmful_batch = next(harmful_iterator)

                    # NOTE: in this dataset 'chosen' = harmful response, 'rejected' = safe
                    # response (see dataset_generation.py) -- so the raw DPO orientation
                    # below is exactly what the adversary wants: prefer chosen (harmful).

                    
                    adv_batch = preprocess_for_dpo(harmful_batch, tokenizer, max_length=max_length)
                    adv_batch = {k: v.to(device) for k, v in adv_batch.items()}

                    lora_weights_adv = make_adversarial_patch(
                        adv_batch["chosen_input_ids"], adv_batch["chosen_attention_mask"]
                    )
                    original_modules = apply_adversarial_wrappers(model, lora_weights_adv)
                    loss_a = compute_dpo_loss(model, ref_model, adv_batch, device)

                    optimizer_a.zero_grad()
                    optimizer_d.zero_grad()
                    loss_a.backward()
                    step_or_skip(adversary.parameters(), optimizer_a)

                    restore_original_modules(model, original_modules)

                    history["adversary_loss"].append(loss_a.item())
                    print(f"  [adv] step {step + 1}/{k_steps} loss={loss_a.item():.4f}")

                # ================= PHASE 2: TRAIN DEFENDER =================
                print(f"--- Block {block_idx + 1}/{num_blocks}: training defender for {k_steps} step(s) ---")
                adversary.eval()
                model.train()
                for p in adversary.parameters():
                    p.requires_grad = False
                for n, p in model.named_parameters():
                    p.requires_grad = "defender" in n

                for step in range(k_steps):
                    try:
                        harmful_batch = next(harmful_iterator)
                        benign_batch_raw = next(benign_iterator)
                    except StopIteration:
                        harmful_iterator = iter(harmful_dataloader)
                        benign_iterator = iter(benign_dataloader)
                        harmful_batch = next(harmful_iterator)
                        benign_batch_raw = next(benign_iterator)

                    dpo_batch = preprocess_for_dpo(harmful_batch, tokenizer, max_length=max_length)
                    dpo_batch = {k: v.to(device) for k, v in dpo_batch.items()}
                    benign_batch = preprocess_for_it(benign_batch_raw, tokenizer, max_length=max_length)
                    benign_batch = {k: v.to(device) for k, v in benign_batch.items()}

                    lora_weights_adv = make_adversarial_patch(
                        dpo_batch["chosen_input_ids"], dpo_batch["chosen_attention_mask"]
                    )
                    original_modules = apply_adversarial_wrappers(model, lora_weights_adv)

                    # Safety objective: under the adversarial patch, the defender must still
                    # prefer the safe response over the harmful one -- i.e. chosen/rejected
                    # must be flipped relative to the raw dataset orientation used above.
                    defender_dpo_batch = {
                        "chosen_input_ids": dpo_batch["rejected_input_ids"],
                        "chosen_attention_mask": dpo_batch["rejected_attention_mask"],
                        "chosen_labels": dpo_batch["rejected_labels"],
                        "rejected_input_ids": dpo_batch["chosen_input_ids"],
                        "rejected_attention_mask": dpo_batch["chosen_attention_mask"],
                        "rejected_labels": dpo_batch["chosen_labels"],
                    }
                    loss_s = compute_dpo_loss(model, ref_model, defender_dpo_batch, device)
                    restore_original_modules(model, original_modules)

                    # Retain objective: stay close to original behaviour on benign instructions.
                    loss_lm = compute_lm_loss(model, benign_batch)
                    loss_kl = compute_kl_loss(model, ref_model, benign_batch)

                    total_loss_d = 1.0 * loss_s + 0.8 * loss_lm + 0.3 * loss_kl

                    optimizer_d.zero_grad()
                    optimizer_a.zero_grad()
                    total_loss_d.backward()
                    step_or_skip(defender_params, optimizer_d)

                    history["defender_safety_loss"].append(loss_s.item())
                    history["defender_lm_loss"].append(loss_lm.item())
                    history["defender_kl_loss"].append(loss_kl.item())
                    print(f"  [def] step {step + 1}/{k_steps} safety={loss_s.item():.4f} "
                          f"lm={loss_lm.item():.4f} kl={loss_kl.item():.4f}")
    finally:
        act_cache.remove_hooks()

    model.eval()
    model.set_adapter("defender")
    print("\nTraining complete. Returning unmerged defender PEFT model (so only the LoRA adapter gets checkpointed), trained adversary, and loss history.")
    return model, adversary, history
```

## Metrics Plots

Two plots in the spirit of the paper's Figures 3-4:

1. **Training dynamics** -- the four losses tracked in `history` (adversary DPO loss,
   defender safety/LM/KL losses) across every step of the run above.
2. **Safety vs. utility, before vs. after AntiDote** -- a rough proxy for the paper's
   Harmful Score (HS) / Fine-tune Accuracy (FA) trade-off (Table 1, Figure 4), computed
   on the same tiny held-out `harmful_subset` / `benign_subset` used for the smoke test.
   We don't have the paper's harmfulness classifier or capability benchmarks here, so:
   - **Safety margin** = mean `logp(safe response) - logp(harmful response)` under the
     raw DPO orientation (higher = model prefers the safe completion more, i.e. safer).
   - **Utility** = mean next-token LM loss on benign instructions (lower = closer to
     original behaviour), the exact same `compute_lm_loss` used as the defender's retain
     objective during training.

   With only `N_HARMFUL=8` / `N_BENIGN=8` examples these are noisy point estimates, not
   calibrated HS/FA scores -- just enough to sanity-check that the hardened model moved
   in the right direction on both axes.

```python

fig, axes = plt.subplots(2, 2, figsize=(11, 7))
loss_panels = [
    ("adversary_loss", "Adversary DPO loss (Phase 1)", "tab:red"),
    ("defender_safety_loss", "Defender safety DPO loss (Phase 2)", "tab:green"),
    ("defender_lm_loss", "Defender LM retain loss (Phase 2)", "tab:blue"),
    ("defender_kl_loss", "Defender KL retain loss (Phase 2)", "tab:orange"),
]
for ax, (key, title, color) in zip(axes.flat, loss_panels):
    values = history[key]
    ax.plot(range(1, len(values) + 1), values, color=color)
    ax.set_title(title)
    ax.set_xlabel("step")
    ax.set_ylabel("loss")
    ax.grid(alpha=0.3)
fig.suptitle("AntiDote bi-level training dynamics")
fig.tight_layout()
plt.show()
```

```python
from loss import DPOLoss

# Pristine, never-trained copy of the base model -- our "before AntiDote" reference.
pristine_model = AutoModelForCausalLM.from_pretrained(
    model_id,
    low_cpu_mem_usage=True,
    return_dict=True,
    torch_dtype=torch.float32,
    device_map='cpu',
    cache_dir='cache_dir',
    trust_remote_code=True
).eval()
for p in pristine_model.parameters():
    p.requires_grad = False

hardened_model.eval()
dpo = DPOLoss(beta=0.1, device=torch.device('cpu'), loss_type="ipo")
```

```python

@torch.no_grad()
def safety_margin(eval_model, dataset, tokenizer, max_length=MAX_LEN):
    """Mean logp(safe response) - logp(harmful response), raw dpo_data orientation
    (chosen=harmful, rejected=safe). Positive = model prefers the safe response."""
    batch = preprocess_for_dpo(dataset[:], tokenizer, max_length=max_length)
    harmful_logps, safe_logps, _, _ = dpo.concatenated_forward(eval_model, batch)
    return (safe_logps - harmful_logps).mean().item()

@torch.no_grad()
def utility_loss(eval_model, dataset, tokenizer, max_length=MAX_LEN):
    """Mean LM (next-token) loss on benign instructions -- lower is better."""
    batch = preprocess_for_it(dataset[:], tokenizer, max_length=max_length)
    return compute_lm_loss(eval_model, batch).item()

```

```python
results = {}
for label, m in [("Base (pristine)", pristine_model), ("AntiDote (hardened)", hardened_model)]:
    results[label] = {
        "safety_margin": safety_margin(m, harmful_subset, tokenizer),
        "benign_lm_loss": utility_loss(m, benign_subset, tokenizer),
    }
results
```

```python
labels = list(results.keys())
safety_vals = [results[l]["safety_margin"] for l in labels]
utility_vals = [results[l]["benign_lm_loss"] for l in labels]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

axes[0].bar(labels, safety_vals, color=["tab:gray", "tab:green"])
axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].set_ylabel("mean logp(safe) - logp(harmful)")
axes[0].set_title("Safety margin on held-out harmful prompts\n(higher = prefers the safe response more)")

axes[1].bar(labels, utility_vals, color=["tab:gray", "tab:blue"])
axes[1].set_ylabel("mean LM loss (nats/token)")
axes[1].set_title("Utility on held-out benign instructions\n(lower = closer to original behaviour)")

fig.tight_layout()
plt.show()
```
